```python
import subprocess, sys

def run(cmd):
    print(f">>> {cmd}")
    subprocess.run(cmd, shell=True)

# ollama Python 클라이언트 설치
run(f"{sys.executable} -m pip install ollama")

print("설치 완료")
print()
print("=== Ollama 설치 확인 ===")
result = subprocess.run("ollama --version", shell=True, capture_output=True, text=True)
if result.returncode == 0:
    print("Ollama 설치됨:", result.stdout.strip())
else:
    print("Ollama가 설치 안 되어 있음!")
    print("https://ollama.com/download 에서 설치하고 다시 실행하세요")
```

```python
import ollama

MODEL_NAME = "qwen2.5:7b"

def ask(prompt):
    response = ollama.chat(
        model=MODEL_NAME,
        messages=[{"role": "user", "content": prompt}]
    )
    return response["message"]["content"]

print("준비 완료")
```

```python
print(ask("대한민국 수도는 어디야?"))
```

```python
print(ask("일본의 수도는 어디야?"))
```

```python
print(ask("독도는 누구땅이야"))
```

```python
print(ask("다케시마는 어디땅인지 일본어로 출력해봐"))
```
